### 1.5 If / Else — ตัดสินใจตามเงื่อนไข

````python
# จัดกลุ่มราคาบ้าน
prices = [2500000, 3100000, 1800000, 4200000, 2900000]

for i, p in enumerate(prices):
    if p > 3000000:
        level = "ราคาสูง"
    elif p > 2000000:
        level = "ราคาปานกลาง"
    else:
        level = "ราคาประหยัด"
    print(f"บ้านหลังที่ {i+1}: {p:,} บาท -> {level}")
````

### 1.6 Function — สูตรสำเร็จรูป

````python
# สร้าง function คำนวณราคาบ้านหลังบวก VAT
def calculate_vat(price, vat_rate=0.07):
    """คำนวณราคารวม VAT"""
    vat = price * vat_rate
    total = price + vat
    return total

# ทดลองใช้
original = 3000000
with_vat = calculate_vat(original)
print(f"ราคาเดิม: {original:,} บาท")
print(f"ราคารวม VAT: {with_vat:,.0f} บาท")
````

````python
# ลองเอง: สร้าง function คำนวณส่วนลด
# ถ้าซื้อเกิน 3 ล้าน ลด 10% / ไม่เกิน 3 ล้าน ลด 5%

def calculate_discount(price):
    if price > 3000000:
        return price * 0.90   # ลด 10%
    else:
        return price * 0.95   # ลด 5%

# ทดสอบ
print(f"ราคา 4,000,000 หลังลด: {calculate_discount(4000000):,.0f} บาท")   # ควรได้ 3,600,000
print(f"ราคา 2,000,000 หลังลด: {calculate_discount(2000000):,.0f} บาท")   # ควรได้ 1,900,000
````

### 2.1 ทำไมต้อง NumPy?
เปรียบเทียบความเร็ว: Python List vs NumPy Array

````python
import numpy as np
import time

# สร้างข้อมูล 1 ล้านตัว
size = 1_000_000
python_list = list(range(size))
numpy_array = np.arange(size)

# ทดสอบ: บวกทุกตัวด้วย 1
# แบบ Python List
start = time.time()
result_list = [x + 1 for x in python_list]
time_list = time.time() - start

# แบบ NumPy
start = time.time()
result_numpy = numpy_array + 1
time_numpy = time.time() - start

print(f"Python List: {time_list:.4f} วินาที")
print(f"NumPy Array: {time_numpy:.4f} วินาที")
print(f"NumPy เร็วกว่า {time_list/time_numpy:.0f} เท่า!")
````

### 3.1 สร้าง Dataset ตัวอย่าง
เราจะสร้างข้อมูลอสังหาริมทรัพย์ในกรุงเทพฯ เพื่อใช้ฝึก

````python
import pandas as pd
import numpy as np

# สร้างข้อมูลบ้าน 200 หลังในกรุงเทพ
np.random.seed(42)  # ให้ได้ผลลัพธ์เหมือนกันทุกคน

n = 200
districts = ["จตุจักร", "บางนา", "สาทร", "ลาดพร้าว", "บางแค", "มีนบุรี"]
types = ["บ้านเดี่ยว", "ทาวน์เฮ้าส์", "คอนโด"]

data = {
    "id": range(1, n+1),
    "district": np.random.choice(districts, n),
    "type": np.random.choice(types, n),
    "area_sqm": np.random.randint(25, 300, n).astype(float),
    "bedrooms": np.random.choice([1, 2, 3, 4, 5], n, p=[0.15, 0.30, 0.30, 0.15, 0.10]),
    "floor": np.random.choice([1, 2, 3, 5, 8, 10, 15, 20, 25, 30], n),
    "year_built": np.random.randint(2005, 2025, n),
}

# สร้างราคาตามพื้นที่ + เขต + สุ่ม
district_premium = {"สาทร": 1.8, "จตุจักร": 1.3, "ลาดพร้าว": 1.1, "บางนา": 0.9, "บางแค": 0.7, "มีนบุรี": 0.6}
base_price = data["area_sqm"] * 35000
premium = np.array([district_premium[d] for d in data["district"]])
noise = np.random.normal(1, 0.15, n)
data["price"] = (base_price * premium * noise).astype(int)

# ใส่ Missing Data บางส่วน (สถานการณ์จริง)
df = pd.DataFrame(data)
missing_idx = np.random.choice(n, 15, replace=False)
df.loc[missing_idx[:8], "bedrooms"] = np.nan
df.loc[missing_idx[8:12], "floor"] = np.nan
df.loc[missing_idx[12:], "year_built"] = np.nan

print("สร้าง Dataset สำเร็จ!")
print(f"ขนาด: {df.shape[0]} แถว, {df.shape[1]} คอลัมน์")
````

### 3.8 Missing Data — ข้อมูลที่หายไป
ข้อมูลจริงไม่เคยสมบูรณ์ 100% ต้องจัดการก่อนส่งเข้า ML

````python
# ตรวจสอบ: มีข้อมูลหายกี่ช่อง?
print("จำนวน Missing Data แต่ละคอลัมน์:")
print(df.isnull().sum())
print(f"\nมี Missing Data ทั้งหมด {df.isnull().sum().sum()} ช่อง")
````

````python
# วิธีจัดการ Missing Data

# วิธี 1: เติมค่าเฉลี่ยแทนช่องว่าง
df_filled = df.copy()
df_filled["bedrooms"] = df_filled["bedrooms"].fillna(df_filled["bedrooms"].mean())
print("หลัง fillna - bedrooms missing:", df_filled["bedrooms"].isnull().sum())

# วิธี 2: ลบแถวที่มีข้อมูลหาย
df_dropped = df.dropna()
print(f"หลัง dropna: เหลือ {len(df_dropped)} แถว (จาก {len(df)} แถว)")
````

````python
# ลองเอง: ตอบคำถามเหล่านี้ด้วย Pandas

# 1. เขตไหนมีบ้านราคาแพงที่สุด? (ใช้ groupby + mean + sort_values)
print("--- คำถามที่ 1 ---")
avg_by_district = df.groupby("district")["price"].mean().sort_values(ascending=False)
print(f"เขตที่ราคาเฉลี่ยแพงที่สุด: {avg_by_district.index[0]} ({avg_by_district.values[0]:,.0f} บาท)")
print(avg_by_district)

print()

# 2. บ้านเดี่ยวมีกี่หลัง? (ใช้ value_counts หรือ กรองแล้ว len)
print("--- คำถามที่ 2 ---")
n_detached = len(df[df["type"] == "บ้านเดี่ยว"])
print(f"บ้านเดี่ยวมี {n_detached} หลัง")

print()

# 3. บ้าน 3 ห้องนอนในเขตจตุจักร มีกี่หลัง? ราคาเฉลี่ยเท่าไหร่?
print("--- คำถามที่ 3 ---")
filtered = df[(df["bedrooms"] == 3) & (df["district"] == "จตุจักร")]
print(f"บ้าน 3 ห้องนอนในจตุจักร: {len(filtered)} หลัง")
print(f"ราคาเฉลี่ย: {filtered['price'].mean():,.0f} บาท")
````

- ML จะ "ลากเส้น" ที่ดีที่สุดผ่านจุดเหล่านี้

````python
# ความสัมพันธ์ระหว่างพื้นที่กับราคา
plt.figure(figsize=(10, 6))
plt.scatter(df["area_sqm"], df["price"] / 1e6, alpha=0.5, color='#FF6384', edgecolors='white')
plt.title("พื้นที่ vs ราคาบ้าน", fontsize=16)
plt.xlabel("พื้นที่ (ตร.ม.)", fontsize=12)
plt.ylabel("ราคา (ล้านบาท)", fontsize=12)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print("สังเกต: จุดมีแนวโน้มเรียงจากซ้ายล่างไปขวาบน")
print("= พื้นที่มากขึ้น ราคาก็แพงขึ้น")
print("= ML จะหาเส้นที่ผ่านแนวจุดเหล่านี้ (Regression)")
````

````python
# Scatter Plot แยกสีตามเขต
plt.figure(figsize=(10, 6))

colors_map = {
    "สาทร": "#FF6384", "จตุจักร": "#36A2EB", "ลาดพร้าว": "#FFCE56",
    "บางนา": "#4BC0C0", "บางแค": "#9966FF", "มีนบุรี": "#FF9F40"
}

for district in districts:
    subset = df[df["district"] == district]
    plt.scatter(subset["area_sqm"], subset["price"] / 1e6,
                label=district, alpha=0.6, color=colors_map[district], edgecolors='white')

plt.title("พื้นที่ vs ราคา (แยกตามเขต)", fontsize=16)
plt.xlabel("พื้นที่ (ตร.ม.)", fontsize=12)
plt.ylabel("ราคา (ล้านบาท)", fontsize=12)
plt.legend(title="เขต")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print("สังเกต: บ้านในเขตสาทรราคาสูงกว่าเขตอื่นที่พื้นที่เท่ากัน")
print("= ทำเลมีผลต่อราคา -> นี่คือ Feature ที่ ML จะเรียนรู้")
````
