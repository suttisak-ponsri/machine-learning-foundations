### 1.5 Function — สูตรสำเร็จรูป

สร้างครั้งเดียว เรียกใช้ซ้ำกี่ครั้งก็ได้ มี 3 ส่วน:
1. ชื่อ function
2. parameter (ข้อมูลที่รับเข้ามา)
3. return (ค่าที่ส่งกลับ)

````python
# -- Function คำนวณราคารวม --

def calc_total(price, qty):
    """คำนวณราคารวม = ราคา x จำนวน"""
    total = price * qty
    return total

# เรียกใช้
result = calc_total(65, 3)    # ลาเต้ 3 แก้ว
print(f"ลาเต้ 3 แก้ว = {result} บาท")

result2 = calc_total(45, 2)   # เอสเปรสโซ 2 แก้ว
print(f"เอสเปรสโซ 2 แก้ว = {result2} บาท")
print()

# -- Function พร้อม Default Parameter --

def calc_total_with_discount(price, qty, discount=0):
    """คำนวณราคารวม พร้อมส่วนลด"""
    subtotal = price * qty
    total = subtotal * (1 - discount)
    return total

# ไม่ใส่ส่วนลด (ใช้ค่า default = 0)
print(f"ไม่มีส่วนลด: {calc_total_with_discount(65, 3)} บาท")

# ใส่ส่วนลด 10%
print(f"ลด 10%: {calc_total_with_discount(65, 3, 0.10)} บาท")
````

**โจทย์:** สร้างระบบคิดเงินอย่างง่าย

1. สร้าง Dictionary `menu` เก็บเมนู + ราคา (อย่างน้อย 4 เมนู)
2. สร้าง Function `order(menu_name, qty)` ที่:
   - รับ ชื่อเมนู + จำนวนแก้ว
   - คำนวณราคารวม
   - ถ้าสั่ง 3 แก้วขึ้นไป ลด 10%
   - return ราคาสุดท้าย
3. ทดสอบ: สั่ง ลาเต้ 2 แก้ว กับ ลาเต้ 5 แก้ว

**ลองทำเองก่อน** (5 นาที) แล้วค่อยดูเฉลย cell ถัดไป

````python
# ===========================================
# Workshop 1: ลองเขียนเองตรงนี้
# ===========================================

# 1. สร้าง Dictionary menu


# 2. สร้าง Function order(menu_name, qty)


# 3. ทดสอบ
````

````python
# ===========================================
# Workshop 1: เฉลย
# ===========================================

# 1. สร้าง Dictionary menu
menu = {
    "เอสเปรสโซ": 45,
    "ลาเต้": 65,
    "มอคค่า": 70,
    "อเมริกาโน": 50,
    "คาปูชิโน": 60
}

# 2. สร้าง Function
def order(menu_name, qty):
    """คิดเงิน: สั่ง 3 แก้วขึ้นไป ลด 10%"""
    price = menu[menu_name]      # ดึงราคาจาก Dictionary
    total = price * qty

    if qty >= 3:
        total = total * 0.90     # ลด 10%
        print(f"  สั่ง {menu_name} x{qty} = {price*qty} บาท --> ลด 10% เหลือ {total:.0f} บาท")
    else:
        print(f"  สั่ง {menu_name} x{qty} = {total} บาท")

    return total

# 3. ทดสอบ
print("--- ทดสอบระบบคิดเงิน ---")
order("ลาเต้", 2)     # 65 x 2 = 130 (ไม่ลด)
order("ลาเต้", 5)     # 65 x 5 = 325 --> ลด 10% = 292 บาท
order("เอสเปรสโซ", 4) # 45 x 4 = 180 --> ลด 10% = 162 บาท
````

#### NumPy — คำนวณตัวเลขทั้งชุดในคำสั่งเดียว

````python
import numpy as np

# ยอดขายกาแฟเย็น 7 วัน (จำนวนแก้ว)
days = ["จันทร์", "อังคาร", "พุธ", "พฤหัส", "ศุกร์", "เสาร์", "อาทิตย์"]
sales = np.array([120, 98, 145, 110, 135, 180, 210])

print("ยอดขาย 7 วัน:", sales)
print(f"ค่าเฉลี่ย:    {sales.mean():.1f} แก้ว/วัน")
print(f"สูงสุด:       {sales.max()} แก้ว (วัน{days[sales.argmax()]})")
print(f"ต่ำสุด:       {sales.min()} แก้ว (วัน{days[sales.argmin()]})")
print(f"รวมทั้งสัปดาห์: {sales.sum()} แก้ว")
print()

# คำนวณรายได้ (ราคาแก้วละ 65 บาท)
revenue = sales * 65
print("รายได้แต่ละวัน:", revenue)
print(f"รายได้รวม: {revenue.sum():,} บาท")
````

#### Pandas — อ่านและจัดการตารางข้อมูล

````python
import pandas as pd

# สร้างข้อมูลยอดขาย 30 วันของร้านกาแฟ
np.random.seed(42)

n = 30
data = {
    "day": list(range(1, n+1)),
    "temperature": np.random.uniform(27, 40, n).round(1),
    "drink": np.random.choice(["กาแฟเย็น", "กาแฟร้อน", "ชาเย็น", "โกโก้"], n,
                               p=[0.40, 0.25, 0.20, 0.15]),
    "cups_sold": np.random.randint(30, 250, n),
    "price": np.random.choice([45, 50, 55, 60, 65, 70], n)
}

df = pd.DataFrame(data)

# คำนวณรายได้
df["revenue"] = df["cups_sold"] * df["price"]

print(f"ขนาดข้อมูล: {df.shape[0]} แถว, {df.shape[1]} คอลัมน์")
df.head(10)
````

````python
# สรุปสถิติ
df.describe()
````

````python
# กรองข้อมูล: เฉพาะวันที่อุณหภูมิเกิน 35 องศา
hot_days = df[df["temperature"] > 35]
print(f"วันที่อุณหภูมิเกิน 35 องศา: {len(hot_days)} วัน")
hot_days.head()
````

````python
# GroupBy: รายได้รวมแยกตามเครื่องดื่ม
revenue_by_drink = df.groupby("drink")["revenue"].sum().sort_values(ascending=False)
print("รายได้รวมแยกตามเครื่องดื่ม:")
for drink, rev in revenue_by_drink.items():
    print(f"  {drink}: {rev:,} บาท")
````

#### Matplotlib — สร้างกราฟ

````python
import matplotlib.pyplot as plt

# กราฟแท่ง: รายได้แยกตามเครื่องดื่ม
revenue_by_drink = df.groupby("drink")["revenue"].sum().sort_values(ascending=False)

plt.figure(figsize=(8, 4))
plt.bar(revenue_by_drink.index, revenue_by_drink.values, color=["#6F4E37", "#A67B5B", "#C4A882", "#D4C5A9"])
plt.title("รายได้แยกตามเครื่องดื่ม", fontsize=14)
plt.xlabel("เครื่องดื่ม")
plt.ylabel("รายได้ (บาท)")
plt.tight_layout()
plt.show()
````

````python
# Scatter Plot: อุณหภูมิ vs ยอดขาย
plt.figure(figsize=(8, 4))
plt.scatter(df["temperature"], df["cups_sold"], alpha=0.6, color="#6F4E37", edgecolors="white")
plt.title("อุณหภูมิ vs ยอดขาย", fontsize=14)
plt.xlabel("อุณหภูมิ (องศา)")
plt.ylabel("จำนวนแก้ว")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print("สังเกต: ข้อมูลชุดนี้เป็นแบบสุ่ม จึงยังไม่เห็นความสัมพันธ์ชัด")
print("ในช่วงที่ 2 เราจะสร้างข้อมูลที่มีความสัมพันธ์จริง ๆ")
````

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# --- สร้างข้อมูลจำลอง ---
np.random.seed(42)

n_days = 50

# อุณหภูมิสุ่มระหว่าง 25-40 องศา
temperature = np.random.uniform(25, 40, n_days)

# ยอดขาย = สัมพันธ์กับอุณหภูมิ + noise
# สูตรจริง (ที่เราซ่อนไว้): ยอดขาย = 5 * อุณหภูมิ - 60 + noise
iced_coffee_sold = 5 * temperature - 60 + np.random.normal(0, 10, n_days)
iced_coffee_sold = np.clip(iced_coffee_sold, 20, None).round(0)  # ขายอย่างน้อย 20 แก้ว

# ใส่ใน DataFrame
cafe_data = pd.DataFrame({
    "temperature": temperature.round(1),
    "iced_coffee_sold": iced_coffee_sold.astype(int)
})

print(f"ข้อมูลร้านกาแฟ {n_days} วัน:")
print(f"อุณหภูมิ: {cafe_data['temperature'].min():.1f} - {cafe_data['temperature'].max():.1f} องศา")
print(f"ยอดขาย: {cafe_data['iced_coffee_sold'].min()} - {cafe_data['iced_coffee_sold'].max()} แก้ว")
print()
cafe_data.head(10)
````

````python
# --- Scatter Plot: ดูความสัมพันธ์ ---

X = cafe_data["temperature"].values
Y = cafe_data["iced_coffee_sold"].values

plt.figure(figsize=(9, 5))
plt.scatter(X, Y, color="#6F4E37", alpha=0.7, edgecolors="white", s=60)
plt.title("อุณหภูมิ vs ยอดขายกาแฟเย็น", fontsize=14)
plt.xlabel("อุณหภูมิ (องศา)", fontsize=12)
plt.ylabel("ยอดขายกาแฟเย็น (แก้ว)", fontsize=12)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print("สังเกต: มีแนวโน้มชัดเจนว่า ยิ่งร้อน ยิ่งขายดี")
print("ถ้าลากเส้นตรงผ่านข้อมูลนี้ได้ เราจะทำนายยอดขายจากอุณหภูมิได้")
````

### 2.2 ลองทำ ML แบบมือ

เราจะลอง "เดา" สมการเส้นตรง:

**ยอดขาย = a x อุณหภูมิ + b**

- `a` = ความชัน (slope) -- อุณหภูมิเพิ่ม 1 องศา ขายเพิ่มกี่แก้ว
- `b` = จุดตัดแกน Y (intercept) -- ถ้าอุณหภูมิ = 0 จะขายได้เท่าไหร่

ลองเปลี่ยนค่า `a` กับ `b` ด้านล่าง แล้วดูว่าเส้นฟิตข้อมูลไหม

````python
# =============================================================
# ลองเปลี่ยนค่า a กับ b ดู แล้วรัน cell นี้ใหม่
# =============================================================

a = 3     # <-- ลองเปลี่ยนค่านี้ (เช่น 2, 3, 5, 7, 10)
b = -10   # <-- ลองเปลี่ยนค่านี้ (เช่น 0, -20, -50, -100)

# --- คำนวณค่าทำนาย ---
Y_pred = a * X + b

# --- คำนวณ Error ---
errors = Y - Y_pred                              # Error แต่ละจุด
squared_errors = errors ** 2                      # ยกกำลังสอง
loss = squared_errors.mean()                      # MSE (Mean Squared Error)

# --- แสดงผล ---
plt.figure(figsize=(9, 5))
plt.scatter(X, Y, color="#6F4E37", alpha=0.7, edgecolors="white", s=60, label="ข้อมูลจริง")

# วาดเส้นตรง
x_line = np.linspace(X.min(), X.max(), 100)
y_line = a * x_line + b
plt.plot(x_line, y_line, color="red", linewidth=2, label=f"y = {a}x + ({b})")

# วาดเส้น error (จากจุดข้อมูลไปถึงเส้นตรง)
for i in range(len(X)):
    plt.plot([X[i], X[i]], [Y[i], Y_pred[i]], color="gray", alpha=0.3, linewidth=0.8)

plt.title(f"a = {a}, b = {b} | Loss (MSE) = {loss:,.1f}", fontsize=14)
plt.xlabel("อุณหภูมิ (องศา)", fontsize=12)
plt.ylabel("ยอดขายกาแฟเย็น (แก้ว)", fontsize=12)
plt.legend(fontsize=11)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print(f"สมการ: ยอดขาย = {a} x อุณหภูมิ + ({b})")
print(f"Loss (MSE) = {loss:,.1f}")
print(f"  --> Loss ยิ่งน้อย = เส้นยิ่งฟิตข้อมูลดี")
print()
print("ลองเปลี่ยน a, b แล้วรัน cell นี้ใหม่ ดูว่าค่าไหน Loss น้อยสุด")
````

### 2.3 ให้คอมพิวเตอร์หาค่าที่ดีที่สุด (Brute Force)

แทนที่จะลองทีละค่าด้วยมือ ให้คอมพิวเตอร์ลอง **ทุกค่า** เลย

วิธี: วน loop ลอง a ตั้งแต่ 0 ถึง 10 และ b ตั้งแต่ -200 ถึง 0 แล้วเก็บค่า Loss ทุกคู่

````python
# --- Brute Force: ลองทุกค่า a, b ---

best_a = 0
best_b = 0
best_loss = float("inf")   # เริ่มจาก loss สูงมาก

# ลอง a ตั้งแต่ 0 ถึง 10 (ทีละ 0.1)
# ลอง b ตั้งแต่ -200 ถึง 0 (ทีละ 1)
for a in np.arange(0, 10, 0.1):
    for b in np.arange(-200, 0, 1):
        # คำนวณค่าทำนาย
        y_pred = a * X + b

        # คำนวณ Loss (MSE)
        loss = np.mean((Y - y_pred) ** 2)

        # ถ้า Loss น้อยกว่าที่เคยเจอ -> จำค่านี้ไว้
        if loss < best_loss:
            best_a = a
            best_b = b
            best_loss = loss

print(f"ค่าที่ดีที่สุดที่หาได้:")
print(f"  a = {best_a:.1f}")
print(f"  b = {best_b:.0f}")
print(f"  Loss (MSE) = {best_loss:.1f}")
print()
print(f"สมการ: ยอดขาย = {best_a:.1f} x อุณหภูมิ + ({best_b:.0f})")
````

````python
# --- Plot เส้นที่ดีที่สุด ---

Y_pred_best = best_a * X + best_b

plt.figure(figsize=(9, 5))
plt.scatter(X, Y, color="#6F4E37", alpha=0.7, edgecolors="white", s=60, label="ข้อมูลจริง")

x_line = np.linspace(X.min(), X.max(), 100)
y_line = best_a * x_line + best_b
plt.plot(x_line, y_line, color="red", linewidth=2.5,
         label=f"Best: y = {best_a:.1f}x + ({best_b:.0f}) | Loss = {best_loss:.1f}")

plt.title("เส้นที่ดีที่สุด (Brute Force)", fontsize=14)
plt.xlabel("อุณหภูมิ (องศา)", fontsize=12)
plt.ylabel("ยอดขายกาแฟเย็น (แก้ว)", fontsize=12)
plt.legend(fontsize=11)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
````

````python
# --- ลองทำนายจริง ---

def predict_sales(temp):
    prediction = best_a * temp + best_b
    return round(prediction)

print("ทำนายยอดขายกาแฟเย็น:")
print(f"  พรุ่งนี้ 38 องศา --> ขายได้ประมาณ {predict_sales(38)} แก้ว")
print(f"  มะรืน 28 องศา  --> ขายได้ประมาณ {predict_sales(28)} แก้ว")
print(f"  วันหนาว 25 องศา --> ขายได้ประมาณ {predict_sales(25)} แก้ว")
print()
print("นี่คือ 'Inference' ในภาษา ML")
print("= เอาโมเดลที่เทรนแล้วไปทำนายกับข้อมูลใหม่")
````

````python
# --- Vector = ชุดตัวเลข ---
# ที่เราเขียน np.array(...) มาตลอด ก็คือ Vector นั่นเอง

temp_vector = np.array([32, 35, 28, 30, 33])
sales_vector = np.array([100, 120, 75, 85, 110])

print("Vector อุณหภูมิ:", temp_vector)
print("  shape:", temp_vector.shape)   # (5,) = มี 5 ตัว
print()

# --- Matrix = ตาราง 2 มิติ ---
# ก็คือ array ซ้อน array

data_matrix = np.array([
    [32, 100],
    [35, 120],
    [28,  75],
    [30,  85],
    [33, 110]
])

print("Matrix (ตาราง 2 มิติ):")
print(data_matrix)
print("  shape:", data_matrix.shape)   # (5, 2) = 5 แถว 2 คอลัมน์
````

````python
# --- Dot Product ---
# = เอาตัวเลขคูณกันทีละคู่ แล้วเอาผลรวม

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

# Dot Product ด้วยมือ
manual = (1*4) + (2*5) + (3*6)
print(f"Dot Product ด้วยมือ: (1x4) + (2x5) + (3x6) = {manual}")

# Dot Product ด้วย NumPy
result = np.dot(a, b)
print(f"Dot Product ด้วย NumPy: np.dot(a, b) = {result}")
print()

# --- เชื่อมกับ ML ---
# สมการ y = a*x + b ที่เราทำมา
# จริง ๆ แล้ว พอมีหลาย feature มันจะกลายเป็น:
#
#   y = w1*x1 + w2*x2 + w3*x3 + bias
#     = np.dot(weights, features) + bias
#
# ตัวอย่าง: ทำนายยอดขายจากหลาย feature

weights = np.array([5.0, -2.0, 0.5])    # น้ำหนักแต่ละ feature
features = np.array([35.0, 3.0, 1.0])   # อุณหภูมิ=35, ฝนตก=3mm, วันหยุด=1
bias = -60

prediction = np.dot(weights, features) + bias
print("ทำนายยอดขายจากหลาย feature:")
print(f"  weights  = {weights}")
print(f"  features = {features}")
print(f"  y = np.dot(weights, features) + bias")
print(f"  y = ({weights[0]}x{features[0]}) + ({weights[1]}x{features[1]}) + ({weights[2]}x{features[2]}) + ({bias})")
print(f"  y = {prediction:.0f} แก้ว")
````

### 3.2 Loss Function — AI วัดความผิดพลาดยังไง

จากช่วง 2 เราเห็นแล้วว่า "Error = ค่าจริง - ค่าทำนาย"

แต่จะรวม Error ทุกจุดยังไง? มี 2 วิธีหลัก:

| Loss Function | วิธีคำนวณ | ลักษณะ |
|--------------|-----------|--------|
| MSE (Mean Squared Error) | ยกกำลังสอง แล้วหาค่าเฉลี่ย | ลงโทษค่าผิดเยอะ ๆ หนักกว่า |
| MAE (Mean Absolute Error) | ค่าสัมบูรณ์ แล้วหาค่าเฉลี่ย | ให้น้ำหนักเท่ากันทุกจุด |

````python
# --- ทวน: ข้อมูลจากร้านกาแฟ ---

# ค่าจริง vs ค่าทำนาย (จากสมการที่ดีที่สุดในช่วง 2)
Y_true = Y                              # ยอดขายจริง
Y_pred = best_a * X + best_b            # ยอดขายที่ทำนาย

# Error แต่ละจุด
errors = Y_true - Y_pred

print("ตัวอย่าง Error 10 จุดแรก:")
for i in range(10):
    print(f"  จุดที่ {i+1}: จริง={Y_true[i]:.0f}, ทำนาย={Y_pred[i]:.0f}, Error={errors[i]:+.1f}")
````

````python
# --- MSE (Mean Squared Error) ---
# ยกกำลังสอง -> หาค่าเฉลี่ย
# ข้อดี: ลงโทษจุดที่ผิดเยอะ ๆ หนักมาก (เพราะยกกำลังสอง)

mse = np.mean((Y_true - Y_pred) ** 2)
print(f"MSE = {mse:.2f}")
print()

# --- MAE (Mean Absolute Error) ---
# เอาค่าสัมบูรณ์ -> หาค่าเฉลี่ย
# ข้อดี: เข้าใจง่าย แปลเป็นหน่วยได้เลย (เช่น ผิดเฉลี่ย 8 แก้ว)

mae = np.mean(np.abs(Y_true - Y_pred))
print(f"MAE = {mae:.2f}")
print(f"แปลว่า: โมเดลทำนายผิดเฉลี่ย {mae:.1f} แก้ว/วัน")
print()

# --- เปรียบเทียบ ---
print("ทำไม ML ใช้ MSE มากกว่า MAE?")
print("- MSE ลงโทษหนักถ้าทำนายพลาดเยอะ (เช่น ผิด 50 แก้ว โดน x2500)")
print("- MAE ให้น้ำหนักเท่ากัน (ผิด 50 แก้ว โดนแค่ x50)")
print("- MSE ทำให้โมเดลพยายามหลีกเลี่ยงการทำนายผิดแบบ 'ห่างมาก'")
````

### 3.3 Gradient Descent — AI ปรับตัวเองยังไง

จากช่วง 2 เราใช้ Brute Force (ลองทุกค่า) ซึ่งช้ามาก

**Gradient Descent** ฉลาดกว่า: เหมือน "เดินลงเขาตอนปิดตา"

1. ยืนจุดหนึ่งบนเขา (ค่า a เริ่มต้นแบบสุ่ม)
2. สัมผัสด้วยเท้าว่าทางไหนลาดลง (คำนวณ gradient)
3. ก้าวไปทางนั้น (อัพเดตค่า a)
4. ทำซ้ำจนถึงจุดต่ำสุด (Loss น้อยสุด)

คำศัพท์สำคัญ:
- **Learning Rate** = ก้าวเดินใหญ่หรือเล็ก
- **Epoch** = จำนวนรอบที่เดิน

````python
# --- Gradient Descent อย่างง่าย ---
# หาค่า a ที่ดีที่สุดสำหรับ y = a * x
# (ตัด b ออกก่อน เพื่อให้ง่ายที่สุด)

# ข้อมูลตัวอย่างง่าย ๆ
X_simple = np.array([1, 2, 3, 4, 5], dtype=float)
Y_simple = np.array([2, 4, 6, 8, 10], dtype=float)  # ความจริงคือ y = 2x

# เริ่มต้น: เดาค่า a = 0 (ไม่รู้อะไรเลย)
a = 0.0
learning_rate = 0.04    # ขนาดก้าวเดิน
epochs = 20             # จำนวนรอบ

# เก็บ log ไว้ดูทีหลัง
history = []

print(f"เริ่มต้น: a = {a:.4f}")
print(f"Learning Rate = {learning_rate}")
print(f"Epochs = {epochs}")
print("-" * 50)

for epoch in range(epochs):
    # 1. ทำนาย
    y_pred = a * X_simple

    # 2. คำนวณ Loss (MSE)
    loss = np.mean((Y_simple - y_pred) ** 2)

    # 3. คำนวณ Gradient (ทิศทางที่ต้องปรับ a)
    #    gradient = -2/n * sum(x * (y_true - y_pred))
    gradient = -2 * np.mean(X_simple * (Y_simple - y_pred))

    # 4. อัพเดตค่า a (ก้าวลงเขา)
    a = a - learning_rate * gradient

    # เก็บ log
    history.append({"epoch": epoch+1, "a": a, "loss": loss, "gradient": gradient})

    # แสดงผลทุก 2 รอบ
    if (epoch + 1) % 2 == 0 or epoch == 0:
        print(f"  Epoch {epoch+1:2d}: a = {a:.4f} | Loss = {loss:.4f} | Gradient = {gradient:+.4f}")

print("-" * 50)
print(f"ผลลัพธ์สุดท้าย: a = {a:.4f} (ค่าจริง = 2.0000)")
print(f"Loss สุดท้าย = {history[-1]['loss']:.6f}")
````

````python
# --- Plot: ดู Loss ลดลงทีละ step ---

df_history = pd.DataFrame(history)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# กราฟที่ 1: Loss ลดลงแต่ละ Epoch
axes[0].plot(df_history["epoch"], df_history["loss"], marker="o", color="#D32F2F", linewidth=2, markersize=5)
axes[0].set_title("Loss แต่ละ Epoch", fontsize=13)
axes[0].set_xlabel("รอบที่ (Epoch)")
axes[0].set_ylabel("Loss (MSE)")
axes[0].grid(True, alpha=0.3)

# กราฟที่ 2: ค่า a เข้าใกล้ 2.0 เรื่อย ๆ
axes[1].plot(df_history["epoch"], df_history["a"], marker="o", color="#1976D2", linewidth=2, markersize=5)
axes[1].axhline(y=2.0, color="green", linestyle="--", label="ค่าจริง (a=2.0)")
axes[1].set_title("ค่า a แต่ละ Epoch", fontsize=13)
axes[1].set_xlabel("รอบที่ (Epoch)")
axes[1].set_ylabel("a")
axes[1].legend()
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print("สังเกต:")
print("  - กราฟซ้าย: Loss ลดลงเรื่อย ๆ = AI เรียนรู้ดีขึ้นเรื่อย ๆ")
print("  - กราฟขวา: ค่า a เข้าใกล้ 2.0 (ค่าจริง) เรื่อย ๆ")
print("  - นี่คือวิธีที่ AI 'เรียนรู้' จากข้อมูล")
````

### Workshop 4: ลองปรับ Learning Rate

ลองเปลี่ยนค่า `learning_rate` ด้านล่างแล้วรันดู:
- เล็กมาก: `0.001` -- เกิดอะไรขึ้น?
- พอดี: `0.04` -- (ที่เราใช้ด้านบน)
- ใหญ่เกินไป: `0.1` -- เกิดอะไรขึ้น?

````python
# ===========================================
# Workshop 4: ลองเปลี่ยน Learning Rate
# ===========================================

def run_gradient_descent(lr, epochs=30):
    """รัน Gradient Descent แล้ว return ประวัติ"""
    X_s = np.array([1, 2, 3, 4, 5], dtype=float)
    Y_s = np.array([2, 4, 6, 8, 10], dtype=float)

    a = 0.0
    hist = []

    for ep in range(epochs):
        y_pred = a * X_s
        loss = np.mean((Y_s - y_pred) ** 2)
        gradient = -2 * np.mean(X_s * (Y_s - y_pred))
        a = a - lr * gradient
        hist.append({"epoch": ep+1, "a": a, "loss": loss})

    return pd.DataFrame(hist)

# ลอง 3 ค่า Learning Rate
lr_small = run_gradient_descent(0.001)
lr_good  = run_gradient_descent(0.04)
lr_big   = run_gradient_descent(0.1)

# Plot เปรียบเทียบ
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Loss
axes[0].plot(lr_small["epoch"], lr_small["loss"], label="LR=0.001 (เล็กเกิน)", linewidth=2)
axes[0].plot(lr_good["epoch"],  lr_good["loss"],  label="LR=0.04 (พอดี)", linewidth=2)
axes[0].plot(lr_big["epoch"],   lr_big["loss"],   label="LR=0.1 (ใหญ่เกิน)", linewidth=2, linestyle="--")
axes[0].set_title("Loss แต่ละ Epoch", fontsize=13)
axes[0].set_xlabel("รอบที่ (Epoch)")
axes[0].set_ylabel("Loss")
axes[0].set_ylim(0, 50)
axes[0].legend()
axes[0].grid(True, alpha=0.3)

# ค่า a
axes[1].plot(lr_small["epoch"], lr_small["a"], label="LR=0.001 (เล็กเกิน)", linewidth=2)
axes[1].plot(lr_good["epoch"],  lr_good["a"],  label="LR=0.04 (พอดี)", linewidth=2)
axes[1].plot(lr_big["epoch"],   lr_big["a"],   label="LR=0.1 (ใหญ่เกิน)", linewidth=2, linestyle="--")
axes[1].axhline(y=2.0, color="green", linestyle=":", alpha=0.7, label="ค่าจริง a=2.0")
axes[1].set_title("ค่า a แต่ละ Epoch", fontsize=13)
axes[1].set_xlabel("รอบที่ (Epoch)")
axes[1].set_ylabel("a")
axes[1].legend()
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print("สังเกต:")
print("  LR = 0.001 (เล็กเกิน): ค่อย ๆ ลง ช้ามาก ยังไม่ถึงจุดต่ำสุด")
print("  LR = 0.04  (พอดี):     ลงไปถึงจุดต่ำสุดได้เร็ว")
print("  LR = 0.1   (ใหญ่เกิน): กระเด้งไปมา หรือพุ่งออกนอก")
````

### Gradient Descent กับข้อมูลร้านกาแฟจริง

ลองใช้ Gradient Descent หาค่า a, b ให้กับข้อมูลร้านกาแฟ (แทน Brute Force)

````python
# --- Gradient Descent สำหรับ y = a*x + b (ข้อมูลร้านกาแฟ) ---

# ใช้ข้อมูลจากช่วง 2
X_cafe = cafe_data["temperature"].values
Y_cafe = cafe_data["iced_coffee_sold"].values

# เริ่มต้น
a = 0.0
b = 0.0
learning_rate = 0.0001   # เล็กลงเพราะข้อมูลจริงมีค่าใหญ่กว่า
epochs = 1000
n = len(X_cafe)

history_cafe = []

for epoch in range(epochs):
    # 1. ทำนาย
    y_pred = a * X_cafe + b

    # 2. คำนวณ Loss (MSE)
    loss = np.mean((Y_cafe - y_pred) ** 2)

    # 3. คำนวณ Gradient
    grad_a = -2/n * np.sum(X_cafe * (Y_cafe - y_pred))
    grad_b = -2/n * np.sum(Y_cafe - y_pred)

    # 4. อัพเดต
    a = a - learning_rate * grad_a
    b = b - learning_rate * grad_b

    if epoch % 100 == 0:
        history_cafe.append({"epoch": epoch, "a": a, "b": b, "loss": loss})

print(f"Gradient Descent Result:")
print(f"  a = {a:.2f}")
print(f"  b = {b:.2f}")
print(f"  Loss = {loss:.2f}")
print()
print(f"Brute Force Result (จากช่วง 2):")
print(f"  a = {best_a:.1f}")
print(f"  b = {best_b:.0f}")
print(f"  Loss = {best_loss:.2f}")
print()
print("สังเกต: Gradient Descent ได้ค่าใกล้เคียงกับ Brute Force")
print("แต่ไม่ต้องลองทุกค่า ใช้เวลาน้อยกว่ามาก")
````

````python
# --- เปรียบเทียบ Brute Force vs Gradient Descent ---

plt.figure(figsize=(9, 5))
plt.scatter(X_cafe, Y_cafe, color="#6F4E37", alpha=0.7, edgecolors="white", s=60, label="ข้อมูลจริง")

x_line = np.linspace(X_cafe.min(), X_cafe.max(), 100)

# เส้น Brute Force
y_brute = best_a * x_line + best_b
plt.plot(x_line, y_brute, color="blue", linewidth=2, linestyle="--",
         label=f"Brute Force: y={best_a:.1f}x+({best_b:.0f})")

# เส้น Gradient Descent
y_gd = a * x_line + b
plt.plot(x_line, y_gd, color="red", linewidth=2,
         label=f"Gradient Descent: y={a:.2f}x+({b:.1f})")

plt.title("เปรียบเทียบ Brute Force vs Gradient Descent", fontsize=14)
plt.xlabel("อุณหภูมิ (องศา)", fontsize=12)
plt.ylabel("ยอดขายกาแฟเย็น (แก้ว)", fontsize=12)
plt.legend(fontsize=10)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print("ทั้ง 2 เส้นแทบจะทับกัน")
print("แสดงว่า Gradient Descent หาค่าได้ดีเทียบเท่า Brute Force")
print("แต่ Gradient Descent เร็วกว่ามาก (ไม่ต้องลองทุกค่า)")
````
